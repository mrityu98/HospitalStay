```python
cramers_cols = bin_cols + ['gender', 'facid']
```

```python
from scipy.stats import chi2_contingency

def cramers_v(x, y):
    confusion_matrix = pd.crosstab(x, y)

    chi2 = chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()

    phi2 = chi2 / n

    r, k = confusion_matrix.shape

    phi2corr = max(
        0,
        phi2 - ((k - 1) * (r - 1)) / (n - 1)
    )

    rcorr = r - ((r - 1) ** 2) / (n - 1)
    kcorr = k - ((k - 1) ** 2) / (n - 1)

    return np.sqrt(
        phi2corr / max(1, min(kcorr - 1, rcorr - 1))
    )
```

```python
cramers_matrix = pd.DataFrame(
    np.zeros((len(cramers_cols), len(cramers_cols))),
    index=cramers_cols,
    columns=cramers_cols
)

for col1 in cramers_cols:
    for col2 in cramers_cols:
        cramers_matrix.loc[col1, col2] = cramers_v(
            df_train[col1],
            df_train[col2]
        )
```

```python
plt.figure(figsize=(16, 12))

sb.heatmap(
    cramers_matrix,
    annot=True,
    fmt=".2f",
    cmap="YlGnBu",
    vmin=0,
    vmax=1
)

plt.title(
    "Cramer's V Association Heatmap",
    fontsize=16
)
plt.savefig('./Images/cramersV.png',dpi=300)
plt.tight_layout()
```
