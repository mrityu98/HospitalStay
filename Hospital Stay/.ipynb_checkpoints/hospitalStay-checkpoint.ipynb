```python
datetime_cols=['vdate','discharged']
cat_cols=['rcount','gender','facid','comorbidity_count','abnormal_lab_count']
bin_cols=['dialysisrenalendstage','asthma','irondef','pneum','substancedependence','psychologicaldisordermajor','depress','psychother','fibrosisandother','malnutrition','hemo','has_any_condition','flag_psych_cluster','flag_hematological','flag_respiratory','flag_metabolic_risk'] + lab_flags
num_cols=['hematocrit','neutrophils','sodium','glucose','bloodureanitro','creatinine','bmi','pulse','respiration']
target='lengthofstay'
```

```python
df_train = pd.get_dummies(df_train, columns=['facid', 'gender'], drop_first=True, dtype=int)
df_test  = pd.get_dummies(df_test,  columns=['facid', 'gender'], drop_first=True, dtype=int)
```

```python
df_test = df_test.reindex(columns=df_train.columns, fill_value=0)
```

```python
cat_cols.remove('facid')
cat_cols.remove('gender')
```

```python
X_train=df_train.drop(columns=target)
y_train=df_train[target]
```

```python
X_train.shape
```

```python
X_test=df_test.drop(columns=target)
y_test=df_test[target]
```

```python
X_test.shape
```

```python
from sklearn.preprocessing import PowerTransformer

# Fit on train only
pt = PowerTransformer(method='yeo-johnson', standardize=False)

cols_to_transform = ['hematocrit','neutrophils',
            'bloodureanitro','respiration']

X_train[cols_to_transform] = pt.fit_transform(X_train[cols_to_transform])
X_test[cols_to_transform]  = pt.transform(X_test[cols_to_transform])  
```

```python
from sklearn.feature_selection import VarianceThreshold

vt = VarianceThreshold(threshold=0.01)
vt.fit(X_train[cat_cols+bin_cols])
kept = X_train[cat_cols+bin_cols].columns[vt.get_support()]
print(f"Dropped {X_train[cat_cols+bin_cols].shape[1] - len(kept)} near-constant features")
```

```python
print(kept)
print(cat_cols+bin_cols)
```

```python
categ_cols=bin_cols+cat_cols
for col in categ_cols:
    if col not in kept:
        X_train.drop(columns=col,inplace=True)

print(X_train.shape)
```

```python
from sklearn.feature_selection import SelectKBest, f_regression

# f_regression computes F-statistic between each feature and target
X_train_vt=X_train.copy()
selector = SelectKBest(score_func=f_regression, k='all')
selector.fit(X_train_vt, y_train)

# Build a ranked score dataframe
f_scores = pd.DataFrame({
    'feature': X_train_vt.columns,
    'f_score': selector.scores_,
    'p_value': selector.pvalues_
}).sort_values('f_score', ascending=False)

# Drop statistically insignificant features (p > 0.05)
significant = f_scores[f_scores['p_value'] < 0.05]['feature'].tolist()
X_train_filtered = X_train_vt[significant]
print(f"Kept {len(significant)} statistically significant features")
```

```python
plt.figure(figsize=(16,14))

sb.barplot(
    data=f_scores,
    x='f_score',
    y='feature'
)

plt.title('F regression Scores')
plt.xlabel('f_scores')
plt.ylabel('Feature')

plt.tight_layout()

# Save as PNG
plt.savefig(
    './Images/f_scores.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant
import numpy as np

def compute_vif(X):
    X_const = add_constant(X)
    vif = pd.DataFrame({
        'feature': X_const.columns,
        'VIF': [variance_inflation_factor(X_const.values, i)
                for i in range(X_const.shape[1])]
    })
    return vif[vif['feature'] != 'const'].sort_values('VIF', ascending=False)

# Iterative VIF — drop highest VIF feature one at a time until all < 10
# This is the correct industry approach, not dropping all > 10 at once

X_vif = X_train_filtered.copy()

while True:
    vif_df = compute_vif(X_vif)
    max_vif = vif_df['VIF'].max()
    if max_vif < 10:
        break
    worst_feature = vif_df.iloc[0]['feature']
    print(f"Dropping '{worst_feature}' with VIF={max_vif:.2f}")
    X_vif = X_vif.drop(columns=[worst_feature])

print(f"\nFinal features after VIF: {X_vif.shape[1]}")
print(X_vif.columns.tolist())
```
