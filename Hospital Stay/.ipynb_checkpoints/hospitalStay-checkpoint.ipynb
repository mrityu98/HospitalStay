```python
from sklearn.feature_selection import SelectKBest, f_regression

# f_regression computes F-statistic between each feature and target
X_train_vt=X_train.copy()
selector = SelectKBest(score_func=f_regression, k='all')
selector.fit(X_train_vt, y_train)

# Build a ranked score dataframe
f_scores = pd.DataFrame({
    'feature': X_train_vt.columns,
    'f_score': selector.scores_,
    'p_value': selector.pvalues_
}).sort_values('f_score', ascending=False)

# Drop statistically insignificant features (p > 0.05)
significant = f_scores[f_scores['p_value'] < 0.05]['feature'].tolist()
X_train_filtered = X_train_vt[significant]
print(f"Kept {len(significant)} statistically significant features")
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant
import numpy as np

def compute_vif(X):
    X_const = add_constant(X)
    vif = pd.DataFrame({
        'feature': X_const.columns,
        'VIF': [variance_inflation_factor(X_const.values, i)
                for i in range(X_const.shape[1])]
    })
    return vif[vif['feature'] != 'const'].sort_values('VIF', ascending=False)

# Iterative VIF — drop highest VIF feature one at a time until all < 10
# This is the correct industry approach, not dropping all > 10 at once

X_vif = X_train_filtered.copy()

while True:
    vif_df = compute_vif(X_vif)
    max_vif = vif_df['VIF'].max()
    if max_vif < 10:
        break
    worst_feature = vif_df.iloc[0]['feature']
    print(f"Dropping '{worst_feature}' with VIF={max_vif:.2f}")
    X_vif = X_vif.drop(columns=[worst_feature])

print(f"\nFinal features after VIF: {X_vif.shape[1]}")
print(X_vif.columns.tolist())
```

```python
cols_to_scale=[col for col in X_vif.columns if X_vif[col].dtype in ['category','float64']]
cols_to_scale.append('pulse')
cols_to_scale
```

```python
from sklearn.linear_model import LassoCV
from sklearn.preprocessing import StandardScaler

# Scale first — mandatory before Lasso
X_scaled=X_vif.copy()
scaler = StandardScaler()
X_scaled[cols_to_scale] = scaler.fit_transform(X_scaled[cols_to_scale])
#X_scaled_df = pd.DataFrame(X_scaled, columns=X_scaled_df.columns)

# LassoCV finds optimal alpha via cross-validation automatically
lasso_cv = LassoCV(cv=5, random_state=42, max_iter=10000)
lasso_cv.fit(X_scaled, y_train)

# Final selected features = non-zero coefficients
coef_df = pd.DataFrame({
    'feature': X_vif.columns,
    'coefficient': lasso_cv.coef_
}).sort_values('coefficient', key=abs, ascending=False)

final_features = coef_df[coef_df['coefficient'] != 0]['feature'].tolist()
print(f"Lasso selected {len(final_features)} features:")
print(coef_df[coef_df['coefficient'] != 0].to_string(index=False))
```

```python
print(f"LassoCV best alpha : {lasso_cv.alpha_:.6f}")
```

```python
from sklearn.metrics import (
    mean_absolute_error, mean_squared_error, r2_score
)

def wmape(y_true, y_pred):
    """Weighted Mean Absolute Percentage Error — handles zero actuals safely"""
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    return np.sum(np.abs(y_true - y_pred)) / np.sum(np.abs(y_true))
 
def adjusted_r2(y_true, y_pred, n_features):
    """Adjusted R² accounting for number of predictors"""
    n = len(y_true)
    r2 = r2_score(y_true, y_pred)
    return 1 - (1 - r2) * (n - 1) / (n - n_features - 1)
 
def compute_metrics(y_true, y_pred, n_features, prefix=''):
    """Returns dict of all 4 metrics"""
    return {
        f'{prefix}MAE'         : mean_absolute_error(y_true, y_pred),
        f'{prefix}RMSE'        : np.sqrt(mean_squared_error(y_true, y_pred)),
        f'{prefix}WMAPE'       : wmape(y_true, y_pred),
        f'{prefix}Adj_R2'      : adjusted_r2(y_true, y_pred, n_features),
    }
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import (
    KFold, GridSearchCV
)
def build_selective_scaler(feature_list,cols_to_scale):
    """
    ColumnTransformer that scales only num+count cols.
    Binary/flag/OHE cols pass through unchanged.
    Works correctly inside Pipeline → no leakage across folds.
    """
    passthru_cols = [col for col in feature_list if col not in cols_to_scale]
 
    transformer = ColumnTransformer(transformers=[
        ('scale'   , StandardScaler(), cols_to_scale),
        ('passthru', 'passthrough'   , passthru_cols),
    ])
    return transformer
```

```python
from sklearn.linear_model import (
    LinearRegression, Ridge, Lasso, ElasticNet, LassoCV
)

models = {
    'LinearRegression' : LinearRegression(),
    'Ridge'            : Ridge(alpha=1.0),
    'Lasso'            : Lasso(alpha=lasso_cv.alpha_, max_iter=10000),
    'ElasticNet'       : ElasticNet(alpha=lasso_cv.alpha_,
                                    l1_ratio=0.5, max_iter=10000),
}
 
def build_pipeline(model, feature_list):
    return Pipeline([
        ('scaler', build_selective_scaler(feature_list,cols_to_scale)),
        ('model' , model)
    ])
```

```python
X_cv = X_vif[final_features]
kf   = KFold(n_splits=5, shuffle=True, random_state=42)
 
all_fold_metrics = []   # one row per model per fold
cv_summary       = []   # mean ± std summary per model
 
for model_name, base_model in models.items():
    fold_records = []
 
    for fold_idx, (train_idx, val_idx) in enumerate(kf.split(X_cv), 1):
        X_fold_train = X_cv.iloc[train_idx]
        X_fold_val   = X_cv.iloc[val_idx]
        y_fold_train = y_train.iloc[train_idx]
        y_fold_val   = y_train.iloc[val_idx]
 
        # Build fresh pipeline each fold — scaler fits only on fold train
        pipe = build_pipeline(base_model, final_features)
        pipe.fit(X_fold_train, y_fold_train)
 
        y_pred = pipe.predict(X_fold_val)
        n_feat = len(final_features)
 
        metrics = compute_metrics(y_fold_val, y_pred, n_feat)
        metrics.update({'Model': model_name, 'Fold': fold_idx})
        fold_records.append(metrics)
        all_fold_metrics.append(metrics)
 
        print(f"  {model_name} | Fold {fold_idx} | "
              f"MAE={metrics['MAE']:.4f}  "
              f"RMSE={metrics['RMSE']:.4f}  "
              f"WMAPE={metrics['WMAPE']:.4f}  "
              f"Adj_R2={metrics['Adj_R2']:.4f}")
 
    # Summary across folds
    fold_df = pd.DataFrame(fold_records)
    summary = {'Model': model_name}
    for metric in ['MAE','RMSE','WMAPE','Adj_R2']:
        summary[f'{metric}_mean'] = fold_df[metric].mean().round(4)
        summary[f'{metric}_std']  = fold_df[metric].std().round(4)
    cv_summary.append(summary)
    print()
 
# DataFrames
fold_metrics_df = pd.DataFrame(all_fold_metrics)
cv_summary_df   = pd.DataFrame(cv_summary).sort_values('Adj_R2_mean', ascending=False)
 
print("=" * 65)
print("CV SUMMARY (mean ± std across 5 folds)")
print("=" * 65)
print(cv_summary_df.to_string(index=False))
```

```python
# ─────────────────────────────────────────────────────────────
# STEP 6 — HYPERPARAMETER TUNING (GridSearchCV)
# ─────────────────────────────────────────────────────────────
# Tuning grids per model
param_grids = {
    'LinearRegression' : {},   # no hyperparameters to tune
    'Ridge'            : {'model__alpha': [0.01, 0.1, 1.0, 10.0, 100.0]},
    'Lasso'            : {'model__alpha': [lasso_cv.alpha_ * f
                                           for f in [0.1, 0.5, 1.0, 2.0, 5.0]]},
    'ElasticNet'       : {'model__alpha'   : [lasso_cv.alpha_ * f
                                              for f in [0.1, 1.0, 5.0]],
                          'model__l1_ratio': [0.2, 0.5, 0.8]},
}
 
best_models    = {}
tuning_results = []
 
for model_name, base_model in models.items():
    pipe      = build_pipeline(base_model, final_features)
    grid      = param_grids[model_name]
 
    if not grid:
        # LinearRegression — no tuning needed, just fit
        pipe.fit(X_cv, y_train)
        best_models[model_name] = pipe
        tuning_results.append({
            'Model'       : model_name,
            'Best_Params' : 'N/A',
            'Best_CV_RMSE': None
        })
        print(f"{model_name}: no hyperparameters to tune")
        continue
 
    gs = GridSearchCV(
        estimator  = pipe,
        param_grid = grid,
        cv         = KFold(n_splits=5, shuffle=True, random_state=42),
        scoring    = 'neg_root_mean_squared_error',
        refit      = True,
        n_jobs     = -1,
        verbose    = 0
    )
    gs.fit(X_cv, y_train)
 
    best_models[model_name] = gs.best_estimator_
    tuning_results.append({
        'Model'       : model_name,
        'Best_Params' : gs.best_params_,
        'Best_CV_RMSE': round(-gs.best_score_, 4)
    })
    print(f"{model_name}: best params = {gs.best_params_}  "
          f"| CV RMSE = {-gs.best_score_:.4f}")
 
tuning_df = pd.DataFrame(tuning_results)
print("\nTuning Summary:")
print(tuning_df.to_string(index=False))
```

```python
# ─────────────────────────────────────────────────────────────
# STEP 7 — RETRAIN ON FULL TRAIN + EVALUATE ON TRAIN & TEST
# ─────────────────────────────────────────────────────────────
X_test_final = X_test[final_features]
final_metrics = []
 
for model_name, fitted_pipe in best_models.items():
    # Retrain on full X_train (best_estimator already refits in GridSearchCV
    # but we make it explicit here for LinearRegression and clarity)
    fitted_pipe.fit(X_cv, y_train)
 
    # Predictions
    y_train_pred = fitted_pipe.predict(X_cv)
    y_test_pred  = fitted_pipe.predict(X_test_final)
 
    n_feat = len(final_features)
 
    train_metrics = compute_metrics(y_train, y_train_pred, n_feat, prefix='Train_')
    test_metrics  = compute_metrics(y_test,  y_test_pred,  n_feat, prefix='Test_')
 
    row = {'Model': model_name}
    row.update(train_metrics)
    row.update(test_metrics)
 
    # Overfitting gap
    row['RMSE_gap']   = round(row['Test_RMSE']  - row['Train_RMSE'],  4)
    row['Adj_R2_gap'] = round(row['Train_Adj_R2'] - row['Test_Adj_R2'], 4)
 
    final_metrics.append(row)
 
    print(f"\n{'='*55}")
    print(f"  {model_name}")
    print(f"{'='*55}")
    print(f"  Train → MAE:{train_metrics['Train_MAE']:.4f}  "
          f"RMSE:{train_metrics['Train_RMSE']:.4f}  "
          f"WMAPE:{train_metrics['Train_WMAPE']:.4f}  "
          f"Adj_R²:{train_metrics['Train_Adj_R2']:.4f}")
    print(f"  Test  → MAE:{test_metrics['Test_MAE']:.4f}  "
          f"RMSE:{test_metrics['Test_RMSE']:.4f}  "
          f"WMAPE:{test_metrics['Test_WMAPE']:.4f}  "
          f"Adj_R²:{test_metrics['Test_Adj_R2']:.4f}")
    print(f"  Overfit gap → RMSE:{row['RMSE_gap']:+.4f}  "
          f"Adj_R²:{row['Adj_R2_gap']:+.4f}")
 
final_metrics_df = pd.DataFrame(final_metrics).sort_values('Test_Adj_R2',
                                                             ascending=False)
```

```python
 # ─────────────────────────────────────────────────────────────
# STEP 8 — SAVE BEST MODEL
# ─────────────────────────────────────────────────────────────
import joblib
 
best_model_name = final_metrics_df.iloc[0]['Model']
best_pipeline   = best_models[best_model_name]
 
joblib.dump(best_pipeline, './Best models/best_linear_model.pkl')
print(f"\nBest model: {best_model_name}")
print(f"Saved to  : ./Best models/best_linear_model.pkl")
 
# Load back (sanity check)
# loaded = joblib.load('best_linear_model.pkl')
# loaded.predict(X_test_final[:5])
```
